```python
import pandas as pd
from lfpr import lfpr_rate

lfpr_df = pd.DataFrame({
    "year": [2000, 2001],
    "Total": [67.0, 67.5],
    "Men": [75.0, 74.8],
    "Women": [60.0, 61.0],
})

# Type
assert isinstance(lfpr_rate(2000, "Total", lfpr_df), float)

# Valid values
assert lfpr_rate(2000, "Men", lfpr_df) == 75.0
assert lfpr_rate(2000, "Women", lfpr_df) == 60.0

# Case-insensitive
assert lfpr_rate(2000, " men ", lfpr_df) == 75.0

# Invariants
men = lfpr_rate(2000, "Men", lfpr_df)
women = lfpr_rate(2000, "Women", lfpr_df)
total = lfpr_rate(2000, "Total", lfpr_df)

assert men != women
assert min(men, women) <= total <= max(men, women)
```
